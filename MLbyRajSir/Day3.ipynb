```python
# OPERATORS, EXPRESSIONS, AND SHORT-CIRCUIT EVALUATION 
# Python operators evaluate variables and values to produce expressions. A professional must understand not just what they output, but how Python evaluates them under the hood to write optimized and bug-free code.
```

```python
# 1. Arithmetic Operators
# These perform standard mathematical calculations. Pay close attention to standard division vs floor division.
```

```python
# Core Arithmetic Demo
num1 = 100
num2 = 7

print(f"Addition (+): {num1 + num2}")
print(f"Subtraction (-): {num1 - num2}")
print(f"Multiplication (*): {num1 * num2}")
print(f"True Division (/): {num1 / num2}   <- Always returns a float")
print(f"Floor Division (//): {num1 // num2}    <- Truncates decimal, returns integer floor")
print(f"Modulus (%): {num1 % num2}     <- Returns the remainder")
print(f"Exponentiation (**):  {5 ** 2}")
```

```python
# 2. Relational (Comparison) Operators
# Relational operators evaluate expressions down to a strict Boolean value (`True` or `False`).
```

```python
val_a = 10
val_b = 12

print(f"Greater Than (>):          {val_a > val_b}")
print(f"Less Than (<):             {val_a < val_b}")
print(f"Greater Than or Equal (>=): {val_a >= 10}")
print(f"Less Than or Equal (<=):    {val_a <= 10}")
print(f"Equality Check (==):       {val_a == val_b}")
print(f"Inequality Check (!=):     {val_a != val_b}")
```

```python
# 3. Assignment & In-Place Operators
# Assignment operators act as efficient shortcuts to modify and update an existing variable's reference in-place.
```

```python
# Initializing baseline variable
score = 15

score += 5   # Equivalent to: score = score + 5
print(f"After += 5:  {score}")

score -= 3   # Equivalent to: score = score - 3
print(f"After -= 3:  {score}")

score *= 2   # Equivalent to: score = score * 2
print(f"After *= 2:  {score}")

score %= 7   # Equivalent to: score = score % 7
print(f"After %= 7:  {score}")
```

```python
# 4. Logical Operators & Short-Circuit Evaluation
# In Python, `and` and `or` do not always return `True` or `False`. Instead, they evaluate truthiness and return the **actual object** that determined the truth value of the expression.
# 
# * **Truthy Values:** Any non-zero number, non-empty string, list, or dictionary.
# * **Falsy Values:** `0`, `0.0`, `""` (empty string), `[]` (empty list), `{}` (empty dict), `None`, `False`.
# 
# **Short-Circuit Rules:**
# * `X and Y`: If `X` is falsy, stops and returns `X`. Otherwise, evaluates and returns `Y`.
# * `X or Y`: If `X` is truthy, stops and returns `X`. Otherwise, evaluates and returns `Y`.
```

```python
# Understanding Truthy/Falsy Outputs
empty_str = ""
active_num = 40

# Logical 'and' evaluation
print("--- Logical AND ---")
print(f"100 and 101:     {100 and 101}")     # Both truthy -> returns the last evaluated value (101)
print(f"empty_str and 10: {empty_str and 10}") # empty_str is falsy -> short-circuits immediately, returns ""
print(f"10 and empty_str: {10 and empty_str}") # 10 is truthy -> moves to empty_str, returns ""

# Logical 'or' evaluation
print("\n--- Logical OR ---")
print(f"10 or 20:        {10 or 20}")        # 10 is truthy -> short-circuits immediately, returns 10
print(f"0 or 20:         {0 or 20}")         # 0 is falsy -> moves to next value, returns 20

# Multiple chaining evaluation
a = "hello"
b = 5
c = []
print("\n--- Chained Short-Circuit ---")
print(f"a and b and c:   {a and b and c}")   # 'a' and 'b' are truthy, hits 'c' (falsy) -> returns []

# Logical 'not' (Always forces a strict boolean output)
print("\n--- Logical NOT ---")
print(f"not 10:          {not 10}")
print(f"not empty_str:   {not empty_str}")
```

```python
# CONDITIONALS AND CONTROL FLOW (IF, ELIF, ELSE)
# 
# Conditional statements allow programs to make decisions and execute specific blocks of code based on whether a condition is truthy or falsy. 
# 
# A professional knows that the order of conditions matters enormously because Python evaluates branches sequentially from top to bottom and exits the block as soon as **one** condition matches.
```

```python
# 1. Structural Blueprint: Standard `if-elif-else` Execution
# Let's observe how structural branches execute. Note that even if multiple conditions could theoretically be true later, only the **first** matching block executes.
```

```python
# Example: Multi-stage check with changing inputs
num1, num2, num3 = 10, 20, 30

if num1 > num2:
    print("Execution hit: 'if' block")
elif num2 > num3:
    print("Execution hit: 'first elif' block")
elif num3 > num1:
    print("Execution hit: 'second elif' block")  # This matches!
else:
    print("Execution hit: 'else' block")
```

```python
# 2. Order of Evaluation Pitfall (Grading Metric Example)
# When testing ranges, always arrange your conditions from the **most restrictive** to the **least restrictive**. 
# 
# If we checked `marks >= 60` first, a student with 98 marks would incorrectly get a "Grade D" because Python would match that first condition and skip the rest.
```

```python
# Production Best Practice for Range Mapping
def calculate_grade(marks: int) -> str:
    """Evaluates and returns structural grade limits based on input scores."""
    if marks >= 95:
        return "Grade A"
    elif marks >= 90:
        return "Grade B"
    elif marks >= 75:
        return "Grade C"
    elif marks >= 60:
        return "Grade D"
    else:
        return "Fail"

# Mock execution tracking
student_score = int(input())
print(f"Score: {student_score} -> Result: {calculate_grade(student_score)}")
```

```python
# 3. Production Solutions for Common Interview/Logic Challenges
# Below are streamlined implementations of baseline control flow questions, formatted cleanly for clear reading.
```

```python
# Challenge 1: Number Sign Classification (Positive / Negative / Zero)
def check_sign(number:int) -> str :
    if number > 0 :
        return "Positive"
    elif number<0 : 
        return "Negative"
    else : return "Zero"

# Challenge 2: Parity Check (Even or Odd via Modulus)
def check_parity(number: int) -> str:
    # If a number is perfectly divisible by 2, remainder is 0
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Challenge 3: Threshold Validation (Voting Eligibility Boundary Check)
def verify_voting_eligibility(age: int) -> str:
    if age >= 18:
        return "Eligible to vote"
    else:
        return "Not eligible"

# Running assertions/tests to display outputs cleanly
print(f"Sign Test (-122): {check_sign(-122)}")
print(f"Parity Test (3):   {check_parity(3)}")
print(f"Voting Test (21):  {verify_voting_eligibility(21)}")
```

```python
# MUTABLE DATA STRUCTURES — LISTS AND DICTIONARIES
# 
# While primitive data types (`str`, `int`, etc.) are immutable, Python relies heavily on mutable collection structures to store, manage, and scale real-world data like JSON responses from API payloads.
```

```python
# 1. Lists: Ordered, Indexable, and Mutable Sequences
# Lists maintain sequence order and allow bidirectional lookup via positive (0-indexed from the left) and negative (starting from -1 from the right) indexing.
```

```python
# Initializing a structural collection list
numbers = [10, 20, 30]

# Positive Indexing (Left-to-Right)
print(f"First element (Index 0):  {numbers[0]}")
print(f"Second element (Index 1): {numbers[1]}")

# Negative Indexing (Right-to-Left / Tail Lookup)
print(f"Last element (Index -1):   {numbers[-1]}")
print(f"Second-to-last (Index -2): {numbers[-2]}")

# Overwriting values directly at explicit memory offsets
numbers[0] = 100
numbers[-1] = 200
print(f"Mutated List Result:      {numbers}")
```

```python
# 2. Dictionaries: Key-Value Hash Maps
# Dictionaries utilize hash tables to maintain explicit key-to-value connections. Keys must be unique and immutable data types (like strings or integers), while values can hold any payload structure.
```

```python
# Structured Dictionary Record Lookup
user_profile = {
    "name": "Raju",
    "full name": "Rajendra from Dholakpur",
    "age": 20,
    "users": ["Ram", "Raj"]  # Embedding mutable structures inside dictionaries
}

print("Basic Key Access")
print(f"Accessing ['name']:       {user_profile['name']}")
print(f"Accessing ['full name']:  {user_profile['full name']}")
print(f"Accessing ['age']:        {user_profile['age']}")

# Drilling into sequential lists nested inside keys
print(f"Complete Nest List:       {user_profile['users']}")
print(f"Target Array Offset [0]:  {user_profile['users'] [0]}")
print(f"Target Array Offset [1]:  {user_profile['users'] [1]}")
```

```python
# INPUT AND OUTPUT MANAGEMENT
# 
# A professional program needs to interact with the outside world. This involves taking input 
# from users (which always arrives as a string) and formatting output cleanly for console logs, 
# dashboards, or user interfaces.
```

```python
# 1. Standard Input & The Type-Casting Imperative
# The built-in `input()` function stops program execution and waits for user terminal input.
# 
#  **Crucial Rule:** `input()` *always* returns data as a `str` (string). If you need to perform 
# mathematical operations on it, you must explicitly cast it using functions like `int()` or `float()`.
```

```python
user_raw_age = input("Enter your age: ")

print(f"Captured Value: {user_raw_age} | Type: {type(user_raw_age)}")

# Direct math will fail on a string: user_raw_age + 1 would raise a TypeError!
# Professional approach: Type-cast immediately upon capture or process safely.
age_in_years = int(user_raw_age)
print(f"Age next year: {age_in_years + 1} | Type: {type(age_in_years)}")
```

```python
# 2. Advanced Output Formatting (F-Strings)
# Formatted string literals (f-strings) provide a clean, readable, and highly optimized way to 
# embed expressions and formatting modifications inside string outputs.
```

```python
item_name = "Premium Georgette Saree"
unit_price = 239.50
quantity = 3
total_cost = unit_price * quantity

# Using standard commas vs Modern Professional F-Strings
print("Item:", item_name, "| Total:", total_cost)

# F-strings allow inline calculation evaluation and number rounding/formatting mechanics
print(f"Item: {item_name}")
print(f"Quantity: {quantity} x ₹{unit_price:.2f}")
print(f"Total Billable Invoice Amount: ₹{total_cost:.2f}")
```

```python
# 3. Production Workflow: Summary
# Here is a complete interactive script block combining Input, Explicit Type Conversion, 
# Dictionary record building, and Formatted Output.
```

```python
# 1. Capture and convert interactive user inputs safely
customer_name = input("Enter customer name: ")
ordered_units = int(input("Enter purchase quantity: "))
base_rate = float(input("Enter unit price (₹): "))

# 2. Package dynamic operational variables into a structural data dictionary record
transaction_record = {
    "client": customer_name,
    "qty": ordered_units,
    "rate": base_rate,
    "grand_total": ordered_units * base_rate
}

# 3. Clean, professional output presentation via automated key evaluations
print("\n" + "="*40)
print("   INVOICE RECEIPT ")
print("="*40)
print(f"Customer Name:   {transaction_record['client'].title()}")
print(f"Items Purchased: {transaction_record['qty']} units")
print(f"Unit Rate:       ₹{transaction_record['rate']:.2f}")
print("-"*40)
print(f"TOTAL PAYABLE:   ₹{transaction_record['grand_total']:.2f}")
```
