```python
# 1. User-Defined Functions and Arbitrary Arguments (`*args`)
# Functions break down complex workflows into reusable blocks.
# * `return` sends a computed value back to the caller and instantly terminates the function.
# * `*args` (Arbitrary Arguments) collects any number of positional arguments into a single **Tuple**.
# 
# **Production Security Note:** The built-in `eval()` function executes a string as code. 
# Never use it with unvalidated user input in production, as it poses a massive security injection risk!
```

```python
# Example 1: Positional Arguments & Multiple Return Values
def calculate_metrics(a: int, b: int):
    """Performs direct math operations and returns multiple values unpacked as a tuple."""
    sum_res = a + b
    max_res = a if a >= b else b
    return sum_res, max_res  # Packing values into a tuple

total, highest = calculate_metrics(200, 500)
print(f"Total: {total} | Max: {highest}")


# Example 2: Arbitrary Positional Arguments (*args)
def stu_names(*names):
    """Accepts any number of input names, collected inside a single tuple."""
    print(f"\nRaw tuple created by *names: {names}")
    for name in names:
        print(f"  Processing: {name}")

stu_names("Golu", "Shae", "Nikhar", "Raju")


# Example 3: Dynamic Calculator using *args (For Reference)
def dynamic_calculator(*args):
    # args[0] = num1, args[1] = num2, args[2] = operational operator string
    expression = f"{args[0]} {args[-1]} {args[1]}"
    result = eval(expression)
    print(f"Eval Expression Check: {expression} = {result}")

dynamic_calculator(10, 20, "+")
dynamic_calculator(10, 3, "//")
```

```python
# 2. Defensive Exception Handling (`try`, `except`, `finally`)
# Runtime execution errors (like dividing by zero or mixing types) crash programs. Defensive code isolates risks inside a `try` block, catches specific errors cleanly with `except`, and uses `finally` for mandatory clean-up actions.
```

```python
#Revision from Python Begineers
try:
    number = int(input())
    result = 10/number
except ZeroDivisionError:
    pass
except ValueError:
    pass
else:
    print(f"result = {result}")
finally:
    print("closing connection")
```

'''Error Handling Best Practices

1. Never Use Bare Excepts
Never write a bare "except:". It catches system exits and Ctrl+C inputs, making your program impossible to stop. Always use "except Exception:" at the bare minimum.

2. Be Explicit, Not Greedy
Only catch errors you know how to fix. If you are looking for a missing file, catch "FileNotFoundError", not a general "Exception".

3. Don't Hide Bugs
If you catch an error and just type "pass", you are burying a bomb. Log it, print it, or fix it immediately.

4. Use Finally for Safekeeping
If you open a database connection or a network socket, close it in the "finally" block. If the code crashes halfway through, "finally" guarantees you won't leak memory.
'''

```python
# 3. Anonymous Lambda Functions
# Lambda functions are single-line, throwing away functions without a name. They are best for short math calculations or simple operations passed inside other utilities.
# 
# **Syntax:** `lambda arguments: expression`
```

```python
# 1. Standard Function vs. Lambda Equivalent
def standard_add(x, y):
    return x + y

lambda_add = lambda x, y: x + y

print("--- Lambda Evaluation ---")
print(f"Standard Function Output: {standard_add(10, 20)}")
print(f"Lambda Equivalent Output: {lambda_add(10, 20)}")

# 2. Re-writing our complex calculator using a compact lambda expression
lambda_calc = lambda *a: eval(f"{a[0]} {a[-1]} {a[1]}")
print(f"Lambda Dynamic Calculator Output: {lambda_calc(10, 20, '*')}")
```

# LIST COMPREHENSIONS (SYNTAX & PERFORMANCE)
# 
# **Syntax Blueprint:**
# `new_list = [expression for item in iterable if condition == True]`

```python
# Baseline Data: A list of item prices from an API
raw_prices = [10, 20, 30, 40, 50]

# --- METHOD 1: The Traditional Way (4 Lines of Code) ---
discounted_prices_old = []
for price in raw_prices:
    discounted_prices_old.append(price * 2)

print("Traditional Output:   ", discounted_prices_old)


# --- METHOD 2: The Professional Way (1 Line of Code) ---
# Read it as: "Multiply price by 2, for every price inside raw_prices"
discounted_prices_new = [price * 2 for price in raw_prices]

print("Comprehension Output: ", discounted_prices_new)
```

# Adding Filter Logic (Combining Loops & Conditionals)
# You can append an `if` statement to the absolute end of the comprehension to filter out unwanted data on the fly.

```python
all_scores = [12, 55, 78, 90, 34, 61, 88]

# Goal: Extract only passing scores (>= 60)
# Read it as: "Keep the score, for every score in all_scores, IF score is greater than or equal to 60"
passing_scores = [score for score in all_scores if score >= 60]

print("Filtered Passing Scores:", passing_scores)
```

# Advanced: Handling Inline Transformations (If-Else Conditions)
# If you need an `else` branch to transform data (rather than filter it out), the syntax shifts. 
# The `if-else` block moves to the **front** of the loop expression.

```python
operational_metrics = [10, 15, 20, 25, 30]

# Goal: Tag numbers as "Even" or "Odd" dynamically
# Syntax: [true_expr if condition else false_expr for item in iterable]
tagged_metrics = ["Even" if num % 2 == 0 else "Odd" for num in operational_metrics]

print("Transformed Target Data:")
print(list(zip(operational_metrics, tagged_metrics))) #Lazy Ecaluation or Deferred Execution is used in this line 
```

# 1. Module Importing Variations
# There are three primary ways to import code in Python. A professional selects the approach 
# that keeps code highly readable and avoids naming conflicts.

```python
# Syntax 1: Global Module Import (Highly Recommended)
import math 
print(f"Global import math usage: {math.sqrt(16)}") # Explicitly namespaces functions

# Syntax 2: Selective Function Import (Best for optimization)
from math import pi, ceil
print(f"Selective target function usage: {ceil(pi)}") # Calls function directly without math. prefix

# Syntax 3: Aliasing Module Names (Standard for heavy analytical packages)
import datetime as dt
print(f"Aliased timestamp initialization: {dt.datetime.now().year}")

# > Production Anti-Pattern: Never use 'from module import *'
# It dumps hundreds of hidden variables into your file, overwriting your own code silently!
```

# 2. The 20% Essential Standard Modules (Production Tools)
# Let's see the most used functions from the four essential modules in action.

```python
# --- MODULE 1: math (Core Numeric Computations) ---
import math

val = 4.7
print("--- math module utilities ---")
print(f"Floor (Round down to integer): {math.floor(val)}") # 4
print(f"Ceil  (Round up to integer):   {math.ceil(val)}")  # 5
print(f"Factorial of 5:                {math.factorial(5)}") # 120


# --- MODULE 2: random (Stochastic Simulations & Sampling) ---
import random

print("\n--- random module utilities ---")
# Generate a random float between 0.0 and 1.0
print(f"Random Float (0.0 - 1.0):      {random.random():.4f}")
# Generate a random integer between a inclusive boundary
print(f"Random Roll (Integer 1-6):     {random.randint(1, 6)}")

# Choose a random item from a sequence list
items = ['Silver', 'Gold', 'Diamond', 'Platinum']
print(f"Random Selection Choice:       {random.choice(items)}")


# --- MODULE 3: datetime (Handling Timestamps and Clocks) ---
from datetime import datetime, timedelta

print("\n--- datetime module utilities ---")
now = datetime.now()
print(f"Current System Precision Clock: {now}")
print(f"Formatted String Representation: {now.strftime('%Y-%m-%d %H:%M:%S')}")

# Calculating offsets (e.g., Token expiration windows)
three_days_later = now + timedelta(days=3)
print(f"Time Offset Calculation (+3D):  {three_days_later.date()}")


# --- MODULE 4: os & sys (System Architecture & Paths) ---
import os
import sys

print("\n--- os & sys execution utilities ---")
# Safe folder/directory paths cross-compatible with Windows/Linux/macOS
current_working_dir = os.getcwd()
print(f"Current Execution Workspace Directory: {current_working_dir}")

# Checking system parameters
print(f"Python Runtime Engine Platform OS:     {sys.platform}")
```
