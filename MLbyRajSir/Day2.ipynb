```python
#VARIABLES, MEMORY MANAGEMENT, AND PRIMARY DATA TYPES
# In Python, variables are not boxes that hold data. Instead, they are **labels (references)** pointing to objects stored in memory.
```

```python
# 1. Memory Optimization: Integer & String Caching
# Python optimizes memory through a mechanism called **interning**. For small integers (from -5 to 256) and short strings without spaces, Python points multiple variables to the exact same memory address (`id`) instead of creating duplicate objects.
```

```python
# Variables pointing to identical short strings share the same memory location
message_1 = "Hello"
message_2 = "Hello"

print(f"message_1 value: {message_1} | Address: {id(message_1)}")
print(f"message_2 value: {message_2} | Address: {id(message_2)}")
print(f"Do they point to the same object? {message_1 is message_2}\n")

# Small Integer Caching (-5 to 256)
x = 250
y = 250
print(f"x address: {id(x)}")
print(f"y address: {id(y)}")
print(f"Do they share memory? {x is y}")
```

```python
# 2. Immutability & Reassignment
# Primary data types like `int`, `float`, `str`, and `bool` are **immutable** (cannot be changed in place). When you modify a variable's value, you aren't changing the data at that memory slot; you are forcing the variable to point to a completely new object at a new memory address.
```

```python
# Reassigning an immutable object
user_name = "Vinay"
print(f"Original: {user_name} | Address: {id(user_name)}")

# Changing the value creates a brand new string object in memory
user_name = "Sharma"
print(f"Updated:  {user_name} | Address: {id(user_name)}")
```

```python
# 3. Core Primitive Data Types
# Let's inspect the four foundational data types using `type()` and look at dynamic typing in action.
```

```python
# Integer (int)
age = 25
print(f"age: {age} | Type: {type(age)}")

# Floating Point (float)
salary_coefficient = 85000.50
print(f"salary_coefficient: {salary_coefficient} | Type: {type(salary_coefficient)}")

# String (str)
full_name = "Vinay Sharma"
print(f"full_name: {full_name} | Type: {type(full_name)}")

# Boolean (bool)
is_active_developer = True
print(f"is_active_developer: {is_active_developer} | Type: {type(is_active_developer)}")
```

```python

# **Production Best Practice:** Never use built-in type names like `str`, `int`, `list`, or `dict` as variable names (e.g., `str = "Vinay"`). Doing so shadows the built-in function, which will break your code down the line when you attempt to use `str()` for type conversion!
```
