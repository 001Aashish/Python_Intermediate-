````python
# LOOPS AND ITERATION (FOR & WHILE)
# 
# Iteration allows code blocks to run repeatedly. A professional understands that a `for` loop is built for **finite collection-driven traversal**, while a `while` loop is built for **state-driven or conditional repetition**.
````

````python
# 1. The `for` Loop and `range()` Exploration
# `range(start, stop, step)` generates structured sequences on the fly. 
# Remember: the `stop` boundary is **exclusive** (never reached).
````

````python
# Generates: 0, 1, 2, 3, 4
for i in range(5):
    print(f"Default Range (0-5): {i}")

# Odd numbers up to 10
for i in range(1, 11, 2):
    print(f"Odd Increment: {i}")

# Countdown sequence
for i in range(5, 0, -1):
    print(f"Countdown: {i}")
````

````python
# 2. Collection Traversal: Index Lookup vs. Direct Iteration
# While you *can* iterate by looking up index ranges (`range(len(collection))`), the clean, Pythonic approach is to iterate over the items **directly**.
````

````python
languages = ["Python", "Go", "Rust"]

print("Traditional Index Iteration (C-Style)")
for i in range(len(languages)):
    print(f"Index {i} maps to -> {languages[i]}")

print("\n Modern Pythonic Direct Iteration")
for lang in languages:
    print(f"Direct Stream Element: {lang}")
````

````python
# 3. Iterating Over Dictionaries & Nested Payloads
# When iterating over a dictionary, you can traverse its `.keys()`, `.values()`, or `.items()`.
````

````python
user_record = {"name": "Aanuj", "age": 21, "role": "Engineer"}

for key, val in user_record.items():
    print(f"Key: {key:<6} | Value: {val}")
# :<6 left-aligns the key string inside a fixed width of 6 characters for clean column formatting

print("\n--- Parsing Nested List Payloads ---")
payload = {"users": ["Niku", "Raju"]}
# Direct loop over the inner nested list object
for user in payload["users"]:
    print(f"Extracted User: {user}")
````

````python
# 4. The `while` Loop Structure
# Runs continuously until its control condition flips to `False`. Always make sure to update your condition variables inside the loop to avoid an infinite loop crash!
````

````python
counter = 1
print(id(counter))
while counter <= 3:
    print(f"While Loop Cycle: {counter}")
    counter += 1  # In-place increment changes state condition
    print(id(counter))
#the id inbuilt fun is used to describe that everytime assignment operator is used then it creates a new obj.
````

````python
# ==============================================================================
# STRUCTURAL CHECKLIST: FOR VS WHILE
# ==============================================================================
# for item in sequence:   -> Best for looping through a known, finite sequence (Lists, Dicts, Ranges).
# while condition:        -> Best when looping until a specific event or state change occurs.
# ==============================================================================
````

````python
# ---
# MENTAL MODEL — WHEN AND HOW TO TRIGGER ITERATION
# 
# To think like a professional engineer, your mind needs a clear set of patterns to determine exactly when and how to loop.
# 
# The Mental Triggers: When do I loop?
# Look at your raw data or your goal. If you see any of these three signs, it's time to use iteration:
# 1. **Plural Data:** Your variable represents a collection rather than a single item (e.g., `users`, `prices`, `payload_records`).
# 2. **Repetitive Actions:** You need to perform the exact same task multiple times (e.g., sending an email to 50 subscribers).
# 3. **Uncertain Horizons:** You need to repeat an action until an event occurs, but you don't know exactly when that will be (e.g., waiting for a user to type `"exit"`).
````

````python
# Choosing the Right Loop

# The Golden Rule: Use a for loop when you know how many times or exactly what you are looping through. Use a while loop when you are waiting for a specific situation to change.

# ### Scenario 1: You know the exact headcount beforehand
# * Go with: for loop using range()
# * Explanation: Use this when you need to run a block of code an exact number of times. It handles its own counting and stops automatically.
# * Real-World Use Case: Kicking off an evaluation script to run exactly 5 test rounds.

# ### Scenario 2: You just need to pull data out of a collection
# * Go with: for loop directly on the collection
# * Explanation: This is your bread and butter for handling data. You tell Python to grab every item from the list, one by one, until it hits the end.
# * Real-World Use Case: Grabbing a list of product prices from an API response and processing them one by one.

# ### Scenario 3: You are waiting on a specific event or status change
# * Go with: while loop
# * Explanation: Use this when you have no clue how long the loop needs to run. It keeps running dynamically until a specific condition flips to False.
# * Real-World Use Case: Repeatedly pinging a server until it finally responds with a successful connection.
````

````python
# ---
# # ALGORITHMIC THINKING — PATTERN MAKING & CONTROLS
# 
# Pattern-making problems are the classic way to build your mental model for nested loops and conditional layouts.
# 
# ### How to Analyze a Pattern Problem
# To solve any grid or pattern puzzle, break it down into a coordinate plane with two variables:
# * **Outer Loop (`r`):** Controls the vertical movement (**Rows**). Think of it as moving down page by page.
# * **Inner Loop (`c`):** Controls the horizontal movement (**Columns**). Think of it as writing characters across a single line.
# * **The Matrix Rule:** The inner loop runs completely from start to finish for *every single step* of the outer loop.
````

````python
# Pattern Challenge 1: The Right-Angled Triangle Grid
# **Goal:** Print an increasing triangle grid of stars.
# ```text
# * # * * # * * * # ```
# **The Logic:** At Row 1, we print 1 star. At Row 2, we print 2 stars. This means our column count equals our current row index (`Columns = Row`).
````

````python
rows = int(input())
# Outer loop dictates vertical row steps
for r in range(1, rows + 1):
    # Inner loop dictates horizontal printing on the current line
    for c in range(1, r + 1):
        print("*", end="|")  # end=" " keeps execution on the same line
        # By default, the print() function has a hidden argument set to end="\n"
    print()  # Empty print drops down to the next row line
    ''' 1)end="" (Empty String): Joins outputs together with absolutely no spaces or breaks.
    2)end=", " (Comma and Space): Excellent for formatting list items or loop outputs on a single line.
    3)end=" | " (Custom Separators): Useful for creating custom text dividers or dashboard layouts.'''
````

````python
# Pattern Challenge 2: Grid Filtering using Conditionals
# **Goal:** Print an advanced coordinate grid, but skip printing stars if the row index matches the column index ($r == c$).
# 
# Combining loops with conditionals gives you total control over your data flow.
````

````python
matrix_dimension = int(input())

for r in range(1, matrix_dimension + 1):
    for c in range(1, matrix_dimension + 1):
        # Injecting control logic inside the iteration stream
        if r == c:
            print("[X]", end=" ")  # Visual flag showing where the indexes match
        else:
            print(f"({r},{c})", end=" ")
    print()
````
