````python
# MASTERING LIST METHODS (THE 80/20 RULE)
# 
# Applying the **80/20 rule** to Python lists means focusing deeply on the 20% of methods that handle 80% of real-world production tasks. 
# 
# Virtually all list tasks boil down to four operational categories: **Adding**, **Removing**, **Searching**, and **Ordering**. 
````

````python
# Initialize baseline list
data_stream = [10, 20, 30]

# 1. append() - Add a single element to the end
data_stream.append(40)
print(f"After append(40):{data_stream}")

# 2. extend() - Unpack and add multiple elements to the end
batch_updates = [100, 200, 300]
data_stream.extend(batch_updates)
print(f"After extend(batch):   {data_stream}")

# 3. insert() - Inject an item at a specific index placement
data_stream.insert(1, 15)  # Injects 15 at index position 1
print(f"After insert(1, 15):   {data_stream}")
````

````python
processing_queue = [10, 100, 20, 30, 50, 60]

# 1. pop() without index drops the trailing element
last_item = processing_queue.pop()
print(f"Popped Item: {last_item} | Remaining Queue: {processing_queue}")

# 2. pop(index) drops from an exact location
targeted_item = processing_queue.pop(1)  # Drops index 1 (100)
print(f"Popped Index 1: {targeted_item} | Remaining Queue: {processing_queue}")

# 3. remove(value) deletes by matching the actual value
processing_queue.remove(20)
print(f"After remove(20):     {processing_queue}")

# 4. clear() instantly purges the container
processing_queue.clear()
print(f"After clear():        {processing_queue}")
````

````python
# 3. Searching and Utility Methods: `index()`, `count()`, and `len()`
# * `len(list)`: A built-in function (not a method) that immediately returns the overall size.
# * `count(value)`: Tells you exactly how many times an element shows up in the list.
# * `index(value)`: Finds the index position of the *first* matching value.
````

````python
metrics = [10, 20, 30, 40, 50, 10, 10, 10]

# Global count tracking
print(f"Total list length:{len(metrics)}")
print(f"How many times 10 appears:{metrics.count(10)}")

# Tracking item locations
print(f"Index location of element 40: {metrics.index(40)}")
````

````python
# 4. Ordering Elements: `sort()` vs `reverse()`
# * `sort()`: Re-orders the list elements ascendingly **in-place** (permanently mutates the list). Pass `reverse=True` for descending order.
# * `reverse()`: Reverses the current order of the list elements completely in-place without sorting them.
````

````python
scores = [10, 20, 30, 40, 50, 15, 18, 20, 56, 64]

# 1. In-place Sorting (Ascending)
scores.sort()
print(f"After sort():                {scores}")

# 2. In-place Sorting (Descending)
scores.sort(reverse=True)
print(f"After sort(reverse=True):    {scores}")

# 3. Plain Reversal (Flips the current structure)
raw_items = [10, 20, 30, 5, 2]
raw_items.reverse()
print(f"After pure reverse():        {raw_items}")
````

````python
# 5. Production Warning: Shallow Copies (`copy()`)
# When assigning a list to a new variable (e.g., `list_b = list_a`), Python copies the **reference**, not the actual data. Modifying `list_b` will unexpectedly change `list_a`! Use `.copy()` to create a clean, safe duplicate.
````

````python
# The Reference Bug
original = [1, 2, 3]
bad_copy = original 
bad_copy.append(99)
print(f"Reference Assignment Bug -> Original changed too!: {original}")

# The Solution
original_clean = [1, 2, 3]
safe_copy = original_clean.copy()
safe_copy.append(99)
print(f"Safe Copy Method -> Original stays safe:{original_clean}")
print(f"Safe Copy Method -> New List isolated:{safe_copy}")
````

````python
# MASTERING DICTIONARY METHODS (THE 80/20 RULE)
# 
# Applying the **80/20 rule** to Python dictionaries means focusing on the core methods used to look up values, modify keys safely, and unpack data. 
# 
# Unlike lists, dictionaries use highly optimized hash-table lookups ($O(1)$ complexity).
````

````python
# 1. Safe Reading & Extraction: `get()` vs Direct Bracket Access
# * `d[key]`: Crashes your program with a `KeyError` if the key does not exist.
# * `d.get(key, default)`: Safely searches for a key. If the key is missing, it returns `None` (or a custom default value) instead of crashing.
````

````python
# Baseline configuration payload
config = {"host": "localhost", "port": 8080}

print(f"Direct Lookup ['host']:{config['host']}")

# Safe fallback handling
print(f"Existing Key Lookup:{config.get('port')}")
print(f"Missing Key (Safe fallback): {config.get('timeout')}") # Returns None
print(f"Missing Key (Custom fallback): {config.get('timeout', 30)}") # Returns 30

# Production Warning Demo (Uncomment to see the crash):
# print(config['timeout']) # -> Raises KeyError: 'timeout'
````

````python
# 2. Collection Unpacking views: `keys()`, `values()`, and `items()`
# Dictionaries allow you to isolate and view components separately. These return dynamic "view objects" that update instantly if the dictionary changes.
````

````python
employee = {"id": 101, "name": "Vijay", "role": "Engineer"}

# 1. .keys() -> Isolates all unique identifiers
print(f"Keys View:   {employee.keys()}")

# 2. .values() -> Isolates all raw data payloads
print(f"Values View: {employee.values()}")

# 3. .items() -> Generates a list-like view of (key, value) pairs as tuples
print(f"Items View:  {employee.items()}")
````

````python
# 3. Modifying & Updating Records: `update()` and `setdefault()`
# * `update(other_dict)`: Merges keys and values from another dictionary, overwriting existing values and appending new ones.
# * `setdefault(key, default)`: Returns the value of a key if it exists. If it doesn't, it inserts the key with the specified default value.
````

````python
user_session = {"user": "Roy", "status": "active"}

# 1. .update() -> Batch merge operations
user_session.update({"status": "idle", "ip_address": "192.168.1.1"})
print(f"After update(): {user_session}")

# 2. .setdefault() -> Safe structural initialization
# If key exists, it just returns the existing value without changing anything
current_theme = user_session.setdefault("theme", "dark")
print(f"After setdefault():  {user_session}")
print(f"Returned Theme:      {current_theme}")
````

````python
# 4. Deleting & Purging Data: `pop()`, `popitem()`, and `clear()`
# * `pop(key)`: Deletes the specified key and returns its value. Throws a `KeyError` if missing unless a fallback is provided.
# * `popitem()`: Removes and returns the *last inserted* (key, value) pair as a tuple.
# * `clear()`: Wipes the entire dictionary.
````

````python
cache = {"session_id": "99xZ", "ttl": 3600, "device": "Mobile"}

# 1. .pop(key) -> Target explicitly by key name
removed_ttl = cache.pop("ttl")
print(f"Popped Value: {removed_ttl} | Remaining Cache: {cache}")

# 2. .popitem() -> Pops the last added pair
popped_pair = cache.popitem()
print(f"Popped Pair:  {popped_pair} | Remaining Cache: {cache}")

# 3. .clear() -> Complete structural wipe
cache.clear()
print(f"After clear(): {cache}")
````

````python
# 5. Dictionary Utilities: Membership checking and Copying
# Just like lists, dictionaries support membership testing using the `in` operator (which checks *keys*, not values) and clean replication using `.copy()`.
````

````python
inventory = {"apples": 50, "bananas": 30}

# Membership checking targets keys directly
print(f"Is 'apples' a key? {'apples' in inventory}")
print(f"Is 50 a key? {50 in inventory} <- (False: checks keys only)")

# Safe memory isolation copy
isolated_inventory = inventory.copy()
print(f"Copied {isolated_inventory}")
print(f"Previous inventory {inventory}")
````
