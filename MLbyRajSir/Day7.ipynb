# ### 1. Designing the Instance Blueprint
# We will build a production-style `Product` model tracking independent internal states.
# Note the difference between instance-level properties and class-level shared properties.

```python
class Product:
    # --- Class Attribute (Shared globally across all product instances) ---
    PLATFORM_TAX_RATE = 0.05 

    # --- Constructor Method ---
    def __init__(self, item_id: str, name: str, base_price: float):
        """
        The __init__ method initializes an object's state.
        'self' acts as a dynamic address label, pointing directly to the specific 
        object instance currently being created in memory.
        """
        # --- Instance Attributes (Unique to each independent object) ---
        self.item_id = item_id
        self.name = name
        self.base_price = base_price

    # --- Instance Method ---
    def calculate_invoice_total(self) -> float:
        """Calculates total price by pulling state values dynamically from self."""
        tax_amount = self.base_price * Product.PLATFORM_TAX_RATE
        return self.base_price + tax_amount
```

# ### 2. Instantiating Objects & Inspecting Memory Binding
# Let's instantiate two separate objects. They share the same class blueprint but maintain completely isolated memory allocations (`id()`).

```python
# Creating isolated object instances
product_1 = Product(item_id="P101", name="Georgette Saree", base_price=200.0)
product_2 = Product(item_id="P102", name="Designer Kurtis", base_price=500.0)

print("--- Instance State Inspection ---")
print(f"Product 1 Name: {product_1.name} | Total Invoice: ₹{product_1.calculate_invoice_total():.2f}")
print(f"Product 2 Name: {product_2.name} | Total Invoice: ₹{product_2.calculate_invoice_total():.2f}")

print("\n--- Memory Isolation Proof ---")
# Even though their structures are identical, their memory locations are unique
print(f"Product 1 Memory ID: {id(product_1)}")
print(f"Product 2 Memory ID: {id(product_2)}")
```

# ### 3. Professional Hygiene: `__str__` vs `__repr__` Magic Methods
# If you print an object directly, Python outputs a cryptic memory address like `<__main__.Product object at 0x... >`. 
# Professionals use magic methods to override this behavior:
# * `__str__`: Returns a clean, readable string for end-users (e.g., logging or UI text).
# * `__repr__`: Returns an explicit, unambiguous string for developers debugging code.

```python
class SmartProduct:
    def __init__(self, name: str, price: float):
        self.name = name
        self.price = price

    def __str__(self) -> str:
        """User-friendly display string."""
        return f"Product: {self.name} (₹{self.price})"

    def __repr__(self) -> str:
        """Developer-focused debugging string."""
        return f"SmartProduct(name='{self.name}', price={self.price})"

# Execution check
premium_item = SmartProduct("Cotton Saree", 450.0)

print("--- String Representations ---")
print("Via __str__ (Standard print): ", premium_item)
print("Via __repr__ (Debugger view):  ", repr(premium_item))
```
